- 신입을 생각하는 단위는 낮음 경력직이 아니라면 와서 경험해도 된다라는 마인드로 들어감




```python
def solution(arr1, arr2):
    import json
    import numpy as np

    data1 = np.array(arr1)
    data2 = np.array(arr2)
    
    # data1 = data1.tolist()
    # data2 = data2.tolist()

    data3 = list()
    data4 = list()
    
    for x in range(0,len(data1)):
        data3.append(data1[x] + data2[x])

    for x in data3:
        data4.append(x.tolist())
        
    
    answer = data4
    return answer
```

```python
solution([[23,44],[2,3]], [[33,11],[33,11]])
```

```python
def sumMatrix(A,B):
    answer = [[c + d for c, d in zip(a,b)] for a, b in zip(A,B)]
    return answer
```

```python
sumMatrix([[23,44],[2,3]], [[33,11],[33,11]])
```

```python
import numpy as np

def multipleMatrix(A,B):
    A_np = np.array(A)
    B_np = np.array(B)
    result = A_np @ B_np
    return result.tolist()
```

```python
multipleMatrix([[1, 4], [3, 2], [4, 1]],[[3, 3], [3, 3]])
```
