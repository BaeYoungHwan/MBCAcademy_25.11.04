```python
#while and for문을 활용하여 작업하기
def solution(stonesList, stepMaxCount):
    maxPeople = 0
    zeroCount = 0
    while zeroCount < stepMaxCount:
        zeroCount = 0
        stonesListNow = []
        
        for x in stonesList:
            
            # 0의 갯수를 세서 Step의 수보다 많다면 결과값 도출.
            if zeroCount < stepMaxCount:
                if x == 0:
                    stonesListNow.append(0)
                    zeroCount += 1
                else:
                    stonesListNow.append(x-1)
                    zeroCount = 0
        
            else:
        
                break
    
        if zeroCount < stepMaxCount:
            maxPeople += 1     
            stonesList = stonesListNow
            
    return maxPeople
    
```

```python
def solution(stonesList, stepMaxCount):
    # 탐색할 최소 인원 (1명)과 최대 인원 (가장 튼튼한 돌의 내구도) 설정
    left = 1
    right = max(stonesList)
    answer = 0  # 결과값(건널 수 있는 최대 인원)을 저장할 변수

    # 이진 탐색 시작
    while left <= right:
        # mid = 이번에 시험해볼 인원 수 (중간값)
        mid = (left + right) // 2

        # 현재 mid명이 건널 수 있는지를 판단하기 위한 변수들
        zero_count = 0  # 연속으로 건널 수 없는 돌의 개수 카운트
        can_cross = True  # mid명이 무사히 건널 수 있는지를 나타냄 (True면 가능)

        # 모든 돌을 하나씩 확인
        for stone in stonesList:
            # 돌의 내구도 - 사람이 건너는 횟수(mid)
            # 0 이하가 되면, 그 돌은 밟을 수 없음
            if stone - mid <= 0:
                zero_count += 1  # 밟을 수 없는 돌 1개 추가
                # 만약 연속으로 밟을 수 없는 돌이 stepMaxCount 이상이면 건널 수 없음
                if zero_count >= stepMaxCount:
                    can_cross = False
                    break  # 더 볼 필요 없이 실패
            else:
                # 돌을 밟을 수 있으면 연속 실패 카운트를 초기화
                zero_count = 0

        # mid명이 건널 수 있었다면 → 더 많은 사람도 가능한지 확인
        if can_cross:
            answer = mid          # 현재 인원 수를 가능한 최대값 후보로 저장
            left = mid + 1        # 범위를 오른쪽으로 옮겨서 더 많은 인원 시도
        else:
            # mid명이 못 건너면 → 인원을 줄여서 다시 시도
            right = mid - 1

    # 최종적으로 가능한 최대 인원 수 반환
    return answer
```
