```python
#GPT가 풀어준 DFS 버전
# 권장 DFS 풀이
def solution(n, computers):
    visited = [False] * n
    answer = 0

    def dfs(node):
        visited[node] = True
        for next_node in range(n):
            if computers[node][next_node] == 1 and not visited[next_node]:
                dfs(next_node)

    for i in range(n):
        if not visited[i]:
            dfs(i)
            answer += 1

    return answer
```

```python
#GPT가 풀어준 BFS 버전

from collections import deque

def solution(n, computers):
    visited = [False] * n
    answer = 0
    
    for i in range(n):
        if not visited[i]:
            queue = deque([i])
            visited[i] = True
            
            while queue:
                node = queue.popleft()
                for next_node in range(n):
                    if computers[node][next_node] == 1 and not visited[next_node]:
                        visited[next_node] = True
                        queue.append(next_node)
            
            answer += 1
    
    return answer
```
