```python
#BFS로는 효율성 검사를 통과할 수 없다.
def solution(triangle):
    from collections import deque
    
    triDeque = deque()
    triDeque.append([triangle[0][0], (0,0)])
    sumMax = []
    sumlist = [[0]*x for x in range(1,len(triangle)+1)]
    
    while triDeque:
        
        sumNum, cur = triDeque.popleft()
    
        if cur[0] == len(triangle)-1:
            sumMax.append(sumNum)
            continue

        
        
        sumNum1 = triangle[cur[0]+1][cur[1]] + sumNum
        if sumNum1 > sumlist[cur[0]+1][cur[1]]:
            sumlist[cur[0]+1][cur[1]] = sumNum1
            triDeque.append([sumNum1, (cur[0]+1, cur[1])])
            
        
        sumNum2 = triangle[cur[0]+1][cur[1]+1] + sumNum
        if sumNum2 > sumlist[cur[0]+1][cur[1]+1]:
            sumlist[cur[0]+1][cur[1]+1] = sumNum2
            triDeque.append([sumNum2, (cur[0]+1, cur[1]+1)])
    
                        
    return max(sumMax)
```

```python
#BFS 버전의 주석달기
def solution(triangle):
    from collections import deque
    
    # BFS 탐색을 위한 큐
    # 원소 구조: [현재까지의 누적합, (행, 열)]
    triDeque = deque()
    triDeque.append([triangle[0][0], (0,0)])  # 시작점 (0,0)
    
    # 마지막 줄에서 도달한 최대값들 저장
    sumMax = []
    
    # 각 좌표에 도달했을 때의 "최대 누적 합"을 저장하는 DP 보조 테이블
    # sumlist[row][col] = 해당 좌표까지의 최대 경로 누적값
    sumlist = [[0] * x for x in range(1, len(triangle) + 1)]
    sumlist[0][0] = triangle[0][0]
    
    
    while triDeque:
        
        # BFS: 큐에서 현재 위치 꺼내기
        sumNum, cur = triDeque.popleft()
        r, c = cur  # 읽기 편하도록 분해
        
        # 마지막 행에 도달한 경우 → 결과 후보에 추가
        if r == len(triangle) - 1:
            sumMax.append(sumNum)
            continue
        
        # -----------------------------
        # 1) 아래 왼쪽으로 이동 (r+1, c)
        # -----------------------------
        new_r1, new_c1 = r + 1, c
        newSum1 = sumNum + triangle[new_r1][new_c1]
        
        # 기존에 기록된 값보다 크면 갱신하고 큐에 새로운 경로 추가
        # 즉, 더 작은 경로는 폐기하여 중복 탐색 방지 (Pruning)
        if newSum1 > sumlist[new_r1][new_c1]:
            sumlist[new_r1][new_c1] = newSum1
            triDeque.append([newSum1, (new_r1, new_c1)])
        
        
        # -----------------------------
        # 2) 아래 오른쪽으로 이동 (r+1, c+1)
        # -----------------------------
        new_r2, new_c2 = r + 1, c + 1
        newSum2 = sumNum + triangle[new_r2][new_c2]
        
        # 기존 최고값보다 큰 경우에만 큐에 추가
        if newSum2 > sumlist[new_r2][new_c2]:
            sumlist[new_r2][new_c2] = newSum2
            triDeque.append([newSum2, (new_r2, new_c2)])
    
    # 마지막 행에 도달한 모든 경로 중 최댓값 반환
    return max(sumMax)
```
