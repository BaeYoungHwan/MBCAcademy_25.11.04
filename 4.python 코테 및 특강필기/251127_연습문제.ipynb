```python
def solution(x):
    countZero = 0
    countOne = x.count('1')
    countZero += x.count('0')
    number = bin(countOne)
    countN = 1
    
    while int(number,2) > 1:
        countZero += number.count('0')-1
        countOne = number.count('1')
        number = bin(countOne)
        countN += 1
    
    return [countN, countZero]
```

```python
#GPT의 답변 replace를 활용하기
def solution(s):
    count_zero = 0
    count_transform = 0
    
    while s != "1":
        count_zero += s.count("0")      # 0 개수 세기
        s = s.replace("0", "")           # 0 제거
        s = bin(len(s))[2:]              # 길이를 이진수로 변환
        count_transform += 1             # 변환 횟수 증가
    
    return [count_transform, count_zero]
```

```python
n = 15
matchNumber = 0
for x in range(1,n+1):
    count = 0
    for y in range(x,n+1):
        count += y
        if count == n:
            matchNumber += 1

        elif count > n :
            break;
print(matchNumber)         
```

```python
# GPT의 코드보다 시간복잡도에서 밀린다
def solution(n):
    matchNumber = 0
    for x in range(1,n+1):
        count = 0
        for y in range(x,n+1):
            count += y
            if count == n:
                matchNumber += 1
    
            elif count > n :
                break;
    return matchNumber
```

```python
#GPT를 활용한 풀이

def solution(n):
    count = 0
    k = 1
    
    while k * (k + 1) // 2 <= n:  # 최소합이 n을 넘으면 종료
        temp = n - (k * (k - 1) // 2)
        if temp % k == 0:  # a가 자연수 조건
            count += 1
        k += 1
        
    return count
```

최소직사각형(완전탐색)_프로그래머스86491_Python

https://school.programmers.co.kr/learn/courses/30/lessons/86491

```python
max([[60, 50], [30, 70], [60, 30], [80, 40]])
```

```python
minNumber = 0
maxNumber = 0

for x in [[60, 50], [30, 70], [60, 30], [80, 40]]:
    x.sort()
    minN, maxN = x
    if minNumber < minN:
        minNumber = minN
    if maxNumber < maxN:
        maxNumber = maxN

print(minNumber, maxNumber)
```

```python
def solution(sizes):
    minNumber = 0
    maxNumber = 0

    for x in sizes:
        x.sort()
        minN, maxN = x
        if minNumber < minN:
            minNumber = minN
        if maxNumber < maxN:
            maxNumber = maxN

    return minNumber*maxNumber
```

```python
def solution(numbers):
    sumlist = []
    numbers.sort()
    for x in range(len(numbers)-1):
        for y in range(x+1,len(numbers)):
    
            sumlist.append(numbers[x] + numbers[y])
    result = list(set(sumlist))
    result.sort()
    return result
```

```python
# GPT 정답 set에서 sorted를 사용하면 list로 반환

def solution(numbers):
    result = set()
    
    for i in range(len(numbers)):
        for j in range(i+1, len(numbers)):
            result.add(numbers[i] + numbers[j])
    
    return sorted(result)
```
