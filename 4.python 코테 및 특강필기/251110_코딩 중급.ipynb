```python
def solution(record):
    dic = {}
    resultList = []
    # record를 스플릿 하여 하나씩 실행
    for x in record:            
        rlist = x.split(" " )
        
        if rlist[0] == "Enter":

            dic.update({rlist[1] : rlist[2]})
            
            
        elif rlist[0] == "Leave":
            #resultList.append("{}님이 나갔습니다.".format(dic.get(rlist[1])))
            pass
            
        elif rlist[0] == "Change":
            dic[rlist[1]] = rlist[2]
            #resultList.append("{}님이 들어왔습니다.".format(rlist[2]))

    for x in record:            
        rlist = x.split(" ")
        
        if rlist[0] == "Enter":
            resultList.append("{}님이 들어왔습니다.".format(dic.get(rlist[1])))
            
        elif rlist[0] == "Leave":
            resultList.append("{}님이 나갔습니다.".format(dic.get(rlist[1])))
            
            
        else:
            pass

            
    return resultList
        
        
```

```python
def solution(dirs):
    X, Y = 0, 0
    dic = {(0,0)}
    for x in dirs:
        
        if x == "U":
            if Y < 5:
                dic.add((X,Y+0.5))
                Y += 1
                
            else:
                pass
                
        
        elif x == "D":
            if Y > -5:
                dic.add((X,Y-0.5))
                Y -= 1
                
            else:
                pass
                
        
        elif x == "L":
            if X > -5:
                dic.add((X-0.5,Y))
                X -= 1
                

            else:
                pass
                
        
        elif x == "R":
            if X < 5:
                dic.add((X+0.5,Y))
                X += 1

            else:
                pass
                
    return len(dic)-1
    
    

        
```

```python
def solution(k, tangerine):
    result = 0
    while k > 0:
    
        k = k - tangerine.count(max(tangerine, key = tangerine.count))
        my_list = tangerine
        value_to_remove = max(my_list, key = my_list.count)
        tangerine = list(filter(lambda x: x != value_to_remove, my_list))
                
        result += 1


    return result
```
