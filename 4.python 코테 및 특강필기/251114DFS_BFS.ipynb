```python
#인자에 대해 [+, -]를 각각에 대입해서 결과값을 뽑아내고 그 갯수를 세는 방식은 가능 하지만 너무 많이 돌것 같음.
#그래도 한번 만들어 보기는 하겠음. 결과값은 나올테니깐.
#탐색알고리즘이 아니라 인덱스를 이용해서 그냥 값만 구한 알고리즘. 뿐만아니라 경우의수가 누락되는 경우도 많음.

def solution(numbers, target):
    sumList = []
    
    for x in range(len(numbers)):
        for y in range(len(numbers)):        
            
            numbers[y-x] = numbers[y-x] * -1
            sumList.append(sum(numbers))

    return sumList.count(target)
```

```python
# DFS 방법을 사용하는 알고리즘을 만들어보자

def solution(numbers, target):
    answer = 0
    
    def dfs_number(index, sumNumber):
        nonlocal answer

        if(index == len(numbers)):
            if(target == sumNumber):
                answer += 1

            return

        dfs_number(index+1, sumNumber + numbers[index])
        dfs_number(index+1, sumNumber - numbers[index])

    dfs_number(0,0)
    return answer 
            
```

```python
# BFS 방법을 사용하는 알고리즘을 만들어보자

from collections import deque

def solution(numbers, target):
    # BFS 탐색을 위한 큐 준비
    # 큐에는 (인덱스, 현재까지의 합) 을 저장
    queue = deque()
    
    # 초기 상태: 0번째 인덱스, 합은 0
    queue.append((0, 0))
    answer = 0

    # BFS 시작
    while queue:
        idx, current_sum = queue.popleft()  
        # BFS이므로 왼쪽(가장 먼저 들어온 것)부터 꺼냄 → 레벨 순서 유지

        # 1. 모든 숫자를 다 사용한 경우 (= 리프 노드 도달)
        if idx == len(numbers):
            if current_sum == target:  
                answer += 1  # 정답 찾음
            continue  # 다음 큐 요소 확인
            
        # 2. 다음 레벨에 두 가지 경우를 큐에 넣음 (레벨 확장)
        # 2-1. numbers[idx]를 더하는 경우
        queue.append((idx + 1, current_sum + numbers[idx]))

        # 2-2. numbers[idx]를 빼는 경우
        queue.append((idx + 1, current_sum - numbers[idx]))
        # → BFS의 핵심: 지금 단계의 가능한 모든 확장(+) (-)을 큐에 저장
        print(queue)
    return answer
```

```python
# DFS 방식으로 한번 코드를 짜보자
# DFS 방식은 최단거리 알고리즘에 적합하지가 않다.

def solution(maps):
    start = time.perf_counter()
    passList = [[0,0]]
    loadRoot = 1;
    answerList = []
    mapsAll = []
    # Maps을 boolean으로 표현하여 만듬

    mapsAll = [[True]*len(maps[0]) for _ in range(len(maps))]

    
    def dfs_load(locationNow):
        nonlocal loadRoot
        nonlocal answerList        
        #canGo = False       
        if(locationNow == [len(maps)-1,len(maps[0])-1]):
            answerList.append(loadRoot)
            
            return

        
        #상(-1,0) 하(+1, 0) 좌(0, -1), 우(0, +1)
        #상으로 가는 루트
        if(locationNow[0] - 1 >= 0 ):
            
            locationCopy = locationNow.copy()
            locationCopy[0] = locationNow[0] - 1
            
            
            if (maps[locationNow[0]-1][locationNow[1]] == 1 and mapsAll[locationNow[0]-1][locationNow[1]]):
                #canGo = True
                mapsAll[locationNow[0]-1][locationNow[1]] = False
                passList.append(locationCopy)
                loadRoot += 1
                dfs_load(locationCopy)
                loadRoot -= 1
                passList.pop()
                mapsAll[locationNow[0]-1][locationNow[1]] = True

        #하으로 가는 루트
        if(locationNow[0] < len(maps)):
            
            locationCopy = locationNow.copy()
            locationCopy[0] = locationNow[0] + 1
           
            if (maps[locationNow[0]+1][locationNow[1]] == 1 and mapsAll[locationNow[0]+1][locationNow[1]]):
                #canGo = True
                mapsAll[locationNow[0]+1][locationNow[1]] = False
                passList.append(locationCopy)
                loadRoot += 1
                dfs_load(locationCopy)
                loadRoot -= 1
                passList.pop()
                mapsAll[locationNow[0]+1][locationNow[1]] = True
                
        #좌으로 가는 루트
        if(locationNow[1] - 1 >= 0 ):
            
            locationCopy = locationNow.copy()
            locationCopy[1] = locationNow[1] - 1

                       
            if (maps[locationNow[0]][locationNow[1]-1] == 1 and mapsAll[locationNow[0]][locationNow[1]-1]):
                #canGo = True
                mapsAll[locationNow[0]][locationNow[1]-1] = False
                passList.append(locationCopy)
                loadRoot += 1
                dfs_load(locationCopy)
                loadRoot -= 1
                passList.pop()
                mapsAll[locationNow[0]][locationNow[1]-1] = True

        #우으로 가는 루트
        if(locationNow[1] +1 < len(maps[0])):
            
            locationCopy = locationNow.copy()
            locationCopy[1] = locationNow[1] + 1
            
            if (maps[locationNow[0]][locationNow[1]+1] == 1 and mapsAll[locationNow[0]][locationNow[1]+1]):
                #canGo = True
                mapsAll[locationNow[0]][locationNow[1]+1] = False
                passList.append(locationCopy)
                loadRoot += 1
                dfs_load(locationCopy)
                loadRoot -= 1
                passList.pop()
                mapsAll[locationNow[0]][locationNow[1]+1] = True
                

                
        # if(canGo == False):
        #     return

    dfs_load([0,0])

    if answerList:
        return min(answerList)

    else:
        return -1
```

```python
#GPT의 도움을 받은 BFS로 코드를 만들어보자

from collections import deque
import time 
import datetime

def solution(maps):
    # 맵의 크기
    n, m = len(maps), len(maps[0])

    # 방문 여부 체크용 2차원 배열
    # visited[x][y] == True 면 이미 방문한 것
    visited = [[False]*m for _ in range(n)]

    # BFS 큐: (x좌표, y좌표, 현재까지 거리)
    # 시작점은 (0,0)이고 시작 거리 1
    maplist = deque([(0, 0, 1)])
    
    # 시작 위치 방문 처리
    visited[0][0] = True

    # 상 / 하 / 좌 / 우 이동을 위한 방향 벡터
    # dx, dy는 같은 인덱스끼리 매칭됨
    # 예: dx[0]=-1, dy[0]=0 → 위로 이동
    dx =[-1,1,0,0]
    dy =[0,0,-1,1]
    
    while maplist:
        
        # 큐의 맨 앞 요소를 꺼냄 (BFS이므로 FIFO)
        x, y, distance = maplist.popleft()

        
        # 현재 위치가 목표 지점이면 → dist(거리) 반환
        # BFS는 첫 도착이 최단거리이므로 그대로 return
        if x == n-1 and y == m-1:
            return distance

        # 상하좌우 4방향 탐색
        for z in range(4):
            nx = x + dx[z]
            ny = y + dy[z]

                # 1) 지도 범위 안인지 체크
            if(0 <= nx < n and 0 <= ny < m):
                
                # 2) 해당 칸이 갈 수 있는 길(==1)인지 체크
                # 3) 방문한 적 없는 칸인지 체크
                if( maps[nx][ny] == 1 and not visited[nx][ny]):
                    
                    # 방문 처리 (여기서 미리 방문 처리하는 것이 BFS 핵심)
                    visited[nx][ny] = True
                
                    # 갈 수 있으므로 큐에 넣기
                    # 거리(dist)에 +1 해서 다음 레벨로 이동
                    maplist.append((nx, ny, distance+1))

    return -1
                
        
    
```
