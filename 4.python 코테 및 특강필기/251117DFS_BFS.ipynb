```python
# 여행문제 GPT가 풀이 BFS 형식

from collections import deque

def solution(tickets):
    """
    여행 경로 문제 (Travel Route)
    - 모든 티켓을 한 번씩 사용하여 경로를 완성
    - ICN에서 출발
    - 여러 경로가 있을 경우 사전 순으로 가장 작은 경로 선택
    - BFS 방식으로 구현 (deque 사용)
    """

    # 1️⃣ 티켓을 사전 순으로 정렬
    tickets.sort()  # ['ICN','JFK'], ['JFK','HND'], ['HND','IAD'] 등
                    # 사전 순으로 정렬하면 BFS에서 자연스럽게 최소 경로 선택 가능

    # 2️⃣ BFS용 큐 초기화
    # 큐에는 현재 경로와 남은 티켓 목록을 함께 저장
    # 초기 경로는 ['ICN']에서 시작
    queue = deque()
    queue.append( (['ICN'], tickets) )

    # 3️⃣ BFS 탐색
    while queue:
        # 큐에서 현재 상태 꺼내기
        path, remain_tickets = queue.popleft()

        # 3-1️⃣ 남은 티켓이 없으면 완성된 경로 반환
        if not remain_tickets:
            return path  # 모든 티켓 사용 완료, 최적 경로 반환

        # 3-2️⃣ 현재 위치
        last_airport = path[-1]

        # 3-3️⃣ 남은 티켓 중 현재 위치에서 출발 가능한 티켓 찾기
        for i, ticket in enumerate(remain_tickets):
            start, arrive = ticket
            if start == last_airport:
                # 가능한 티켓을 사용하여 새로운 경로 생성
                new_path = path + [arrive]

                # 남은 티켓에서 사용한 티켓 제거
                new_remain = remain_tickets[:i] + remain_tickets[i+1:]

                # 큐에 추가
                queue.append( (new_path, new_remain) )
                # 사전 순을 위해 tickets 정렬 후 BFS 사용 → 자연스럽게 최적 경로 먼저 탐색
```

```python
# GPT가 만든 BFS 방식 _앞뒤로 붙이기 가능 버전


def solution(tickets):
    """
    여행 경로 문제 (Travel Route) - 직관적 BFS + deque 활용
    """
    # 1️⃣ 티켓 사전 순 정렬
    tickets.sort()
    
    # 2️⃣ BFS 큐 초기화
    # 큐에는 deque 형태의 경로와 남은 티켓 리스트를 저장
    queue = deque()
    queue.append( (deque(['ICN']), tickets) )

    while queue:
        path_deque, remain_tickets = queue.popleft()
        
        # 3️⃣ 모든 티켓 사용 시 경로 완성
        if not remain_tickets:
            return list(path_deque)  # deque -> list 변환하여 반환

        last_airport = path_deque[-1]
        first_airport = path_deque[0]

        # 4️⃣ 남은 티켓을 하나씩 확인
        for i, (start, arrive) in enumerate(remain_tickets):
            # 4-1️⃣ 현재 경로 뒤에 연결 가능
            if start == last_airport:
                new_path = deque(path_deque)       # 현재 경로 복사
                new_path.append(arrive)            # 뒤에 붙이기
                new_remain = remain_tickets[:i] + remain_tickets[i+1:]
                queue.append( (new_path, new_remain) )

            # 4-2️⃣ 현재 경로 앞에 연결 가능 (경로 확장 가능성)
            elif arrive == first_airport:
                new_path = deque(path_deque)
                new_path.appendleft(start)         # 앞에 붙이기
                new_remain = remain_tickets[:i] + remain_tickets[i+1:]
                queue.append( (new_path, new_remain) )
```
