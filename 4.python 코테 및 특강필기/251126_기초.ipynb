버스(PCCE 기출문제 7번)_프로그래머스340201_Python

https://school.programmers.co.kr/learn/courses/30/lessons/340201

```python
def func1(num):
    if 0 > num:
        return 0
    else:
        return num

def func2(num):
    if num > 0:
        return 0
    else:
        return num

def func3(station):
    num = 0
    for people in station:
        if people == "Off":
            num += 1
    return num

def func4(station):
    num = 0
    for people in station:
        if people == "On":
            num += 1
    return num


def solution(seat, passengers):
    num_passenger = 0
    for station in passengers:
        num_passenger += func4(station)
        num_passenger -= func3(station)
    answer = func1(seat - num_passenger)

    return answer
```


최대공약수와 최소공배수(연습문제)_프로그래머스12940_Python

https://school.programmers.co.kr/learn/courses/30/lessons/12940


```python
n = 2
m = 5


listN = []
listM = []
result = []

#약수 구하기. N
for x in range(1,(n//2)+1):
    if n % x == 0:
        listN.append(x)
listN.append(n)

#약수 구하기. M
for x in range(1,(m//2)+1):
    if m % x == 0:
        listM.append(x)
listM.append(m)

#최대 공약수 구하기. 
listM.sort(reverse = True)

for x in listM:
    if x in listN:
        result.append(x)
        break;

#최소공배수 구하기
while len(result)<2:

    for x in range(1,n+1):
        if m*x % n == 0:
            result.append(m*x)
            break
    
print(result)
```

```python
def solution(n, m):
        
    listN = []
    listM = []
    result = []

    #약수 구하기. N
    for x in range(1,(n//2)+1):
        if n % x == 0:
            listN.append(x)
    listN.append(n)
    
    #약수 구하기. M
    for x in range(1,(m//2)+1):
        if m % x == 0:
            listM.append(x)
    listM.append(m)
    
    #최대 공약수 구하기. 
    listM.sort(reverse = True)
    
    for x in listM:
        if x in listN:
            result.append(x)
            break;
    
    #최소공배수 구하기
    while len(result)<2:
    
        for x in range(1,n+1):
            if m*x % n == 0:
                result.append(m*x)
                break
        
    return result
```

```python
#GPT를 활용한 최대공약수와 최소공배수
def solution(n, m):
    # 최대공약수 구하는 함수
    def gcd(a, b):
        while b != 0:
            a, b = b, a % b
        return a
    
    # 최소공배수 구하는 함수
    def lcm(a, b):
        return a * b // gcd(a, b)
    
    return [gcd(n, m), lcm(n, m)]
```

https://school.programmers.co.kr/learn/courses/30/lessons/68935

```python
def solution(n):
    base = ''

    while n > 0:
        n, mod = divmod(n, 3)
        base += str(mod)
    return int(base, 3)
```

```python
def solution(number):
    number.sort()
    count = 0
    for x in range(len(number)-2):
        if number[x] > 0:
            break
        for y in range(x+1,len(number)-1):
            if (number[x]+number[y])*-1 in number[y+1:]:
                count += number[y+1:].count((number[x]+number[y])*-1)
    return count
```

가장 가까운 같은 글자(연습문제)_프로그래머스142086_Python

https://school.programmers.co.kr/learn/courses/30/lessons/142086

```python
s = 'banana'
result = []

for x in range(len(s)):
    if s[x] in s[:x]:
        result.append(x - (s[:x].rfind(s[x])))
    else:
        result.append(-1)

print(result)
```
