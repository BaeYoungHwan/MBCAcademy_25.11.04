```python
def solution(m, n, puddles):
    MOD = 1_000_000_007
    
    # dp[y][x]
    dp = [[0] * m for _ in range(n)]
    
    # 물웅덩이 표시
    for x, y in puddles:
        dp[y-1][x-1] = -1  # 물웅덩이
    
    dp[0][0] = 1  # 시작점
    
    for y in range(n):
        for x in range(m):
            
            # 물웅덩이면 스킵
            if dp[y][x] == -1:
                dp[y][x] = 0
                continue
            
            # 위에서 내려오기
            if y > 0:
                dp[y][x] += dp[y-1][x]
            
            # 왼쪽에서 오기
            if x > 0:
                dp[y][x] += dp[y][x-1]
            
            dp[y][x] %= MOD
    
    return dp[n-1][m-1]
```
