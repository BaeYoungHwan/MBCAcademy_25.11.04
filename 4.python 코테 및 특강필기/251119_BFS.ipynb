```python
def solution(begin, target, words):
    from collections import deque

    if (target not in words):
        return 0

    wordDeque = deque()
    distance = 0
    word = begin
    wordDeque.append((word,distance))
    
    while wordDeque:
        word, distance = wordDeque.popleft()
    
        if word == target:
            return distance
            
        for w in words:
            sameWord = 0
            for x in range(len(word)):
                
                if word[x] == w[x]:
                    sameWord += 1    
    
            if(len(word)-1 == sameWord):
                wordDeque.append((w,distance+1))
    return 0
```
