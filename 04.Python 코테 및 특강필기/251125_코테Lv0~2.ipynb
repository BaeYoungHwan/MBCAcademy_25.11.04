음양더하기_월간코드챌린지시즌2_프로그래머스76501_Python

https://school.programmers.co.kr/learn/courses/30/lessons/76501

```python
def solution(absolutes, signs):
    sumX = 0
    
    for x in range(len(absolutes)):
        
        if signs[x]:
            sumX += absolutes[x]
        else:
            sumX -= absolutes[x]
    return sumX
```

```python
solution([4,7,12],[True,False,True])
```

부족한 금액 계산하기(위클리 챌린지)_프로그래머스82612_Python

https://school.programmers.co.kr/learn/courses/30/lessons/82612

```python
def solution(price, money, count):
    
    sumX = 0
    for x in range(1,count+1):
        sumX += x
    if (price*sumX) - money > 0:
        return (price*sumX) - money
    else:
         return 0
```

물부족(PCCE 기출문제)_프로그래머스340202_Python

https://school.programmers.co.kr/learn/courses/30/lessons/340202


```python
def solution(storage, usage, change):
    total_usage = 0
    for i in range(len(change)):
        usage += usage * change[i]/100
        total_usage += usage
        if total_usage > storage:
            return i
    
    return -1
```

```python
def solution(n):
    answer = [[0] * n for _ in range(n)]

    # 이동 방향: 오른쪽, 아래, 왼쪽, 위
    dx = [0, 1, 0, -1]
    dy = [1, 0, -1, 0]

    x, y, direction = 0, 0, 0  # 시작 위치 + 시작 방향

    for num in range(1, n*n + 1):
        answer[x][y] = num

        nx = x + dx[direction]
        ny = y + dy[direction]

        # 방향을 바꿔야 하는 경우
        if nx < 0 or nx >= n or ny < 0 or ny >= n or answer[nx][ny] != 0:
            direction = (direction + 1) % 4
            nx = x + dx[direction]
            ny = y + dy[direction]

        x, y = nx, ny

    return answer
```
