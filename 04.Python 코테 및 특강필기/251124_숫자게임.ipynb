```python
def solution(listA, listB):
    # A와 B를 정렬한다. 
    # 작은 숫자부터 비교하기 위해 반드시 필요.
    listA.sort()
    listB.sort()
    
    count = 0      # B가 승리한 횟수
    index = 0      # B에서 현재 비교할 위치(pointer)
    
    # A의 숫자를 순서대로 확인
    for x in listA:
        # B에서 x보다 큰 숫자를 찾을 때까지 진행
        while index < len(listB):
            
            if x < listB[index]:
                # x보다 큰 숫자를 찾으면 B가 승리
                count += 1
                index += 1   # 해당 숫자를 사용했으니 포인터 이동
                break         # 다음 A 숫자로 이동
                
            else:
                # x보다 작거나 같으면 이길 수 없으니 다음 숫자 확인
                index += 1
    
    return count
```
