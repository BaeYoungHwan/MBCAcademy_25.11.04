```python
def solution(distance, rock, n):
    
    rock.sort()
    rock.append(distance)
    left = 1
    right = distance
    answer = 0
    while left <= right:
        cot = False
        sumN = 0
        mid = (left + right) // 2
        rockCopy = rock.copy()
        
        while sumN <= n:
            
            if rock[0] < mid :
                rock.pop(0)
                sumN += 1
                continue
            
            else:
                for x in range(1,len(rock)):
                    if (rock[x] - rock[x-1])  < mid:
                        rock.remove(rock[x])
                        sumN += 1
                        break   

                    else:
                        cot = True;
                            
            if cot:
                break
                   
        if sumN > n:
            right = mid-1 #미드를 줄여서 더 작은 거리수 찾기.
            
        elif sumN == n:
            between = rock[0]
            for x in range(1,len(rock)):
                if between > (rock[x] - rock[x-1]):
                    between = (rock[x] - rock[x-1])
            if between > (distance - rock[-1]) :
                between = distance - rock[-1]
            answer = between
            left = mid+1   #더 큰 최솟값 찾기.
            
        else:
            left = mid+1
            
        rock = rockCopy 
        
    return answer
```

```python
#GPT가 내준 정답. 
#prev를 사용하고 제거하는것은 removed만 적용해서 prev의 값은 예전 바위 그대로 놔둠


def solution(distance, rocks, n):
    rocks.sort()
    rocks.append(distance)

    left, right = 1, distance
    answer = 0

    while left <= right:
        mid = (left + right) // 2
        removed = 0
        prev = 0

        for rock in rocks:
            if rock - prev < mid:     # 거리가 mid보다 짧으면 제거
                removed += 1
            else:
                prev = rock

        if removed > n:               # 너무 많이 제거해야 한다 = mid가 너무 큼
            right = mid - 1
        else:                         # mid 만들 수 있음 → 더 큰 값 탐색
            answer = mid
            left = mid + 1

    return answer
```
