```python
def solution(number, k):
    number = list(number)  # 문자열을 리스트로 변경 (수정 가능하게)
    length = len(number)
    remove_count = 0
    idx = 0

    while remove_count < k:
        # 앞자리와 비교
        while idx < len(number) - 1 and remove_count < k:
            if number[idx] < number[idx + 1]:
                number.pop(idx)
                remove_count += 1
                # 한 자리 삭제했으니, 이전 자리부터 다시 비교
                idx = max(idx - 1, 0)
                break
            else:
                idx += 1
        else:
            # 끝까지 갔는데 삭제 안 된 경우 -> 남은 만큼 뒤에서 제거
            while remove_count < k:
                # min(number) 대신, 뒤에서부터 제거 (자리값 손실 방지)
                number.pop()
                remove_count += 1
            break

    return ''.join(number)
```

2 x n 타일링 문제

```python
# 타일링 문제는 결국 동적 계획법에 의거하며 그 점화법은 피보나치 수열이랑 동일함
# 마지막 2개의 블록을 어떻게 넣는가를 생각해보면 
# 1개의 공간으로 놓을경우 1(경우의수) * f(x-1) 
# 2개의 공간으로 놓을경우 1(경우의수) * f(x-2) 가로로 세우기만 가능

def solution(n):
    
    #tile = [1,1] # 초기 타일 배치
    
    if n <= 1:     # 타일이 2이하면 해당 초기 셋팅이니 결과값 도출
        return 1

    a, b = 1, 1
    for x in range(2, n+1):   # 3이상일때 bottom-up 알고리즘 사용
        #tile.append(tile[x-1] + tile[x-2])     #리스트에 지속적인 저장으로 인한 메모리가 지속적으로 늘어남
        #tile.append(tile[-1] + tile.pop(0))    #pop을 사용 하면 리스트를 미는 행위이기 때문에 O(n) 복잡도 발생
        a, b = b % 1000000007, (a + b) % 1000000007 #오버플로우가 나타날 수 있기 때문에 % 1000000007을 사용하여 연산

    
    return b % 1000000007
    
    
    
    
```
