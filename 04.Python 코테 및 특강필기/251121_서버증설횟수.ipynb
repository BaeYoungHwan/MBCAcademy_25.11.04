```python
def solution(players, m, k):
    # 최종 서버 증설 횟수 누적
    count = 0
    
    # 각 시간대별 현재 운용 중인 서버 수를 저장
    # servers[i] : i시점에 운용 중인 서버 수
    servers = [0] * len(players)  # 24시간 기준
    
    # 하루 24시간을 순회
    for x in range(len(players)):
        # 현재 시간대에 필요한 최소 서버 수
        server = players[x] // m  # 몫 연산 (서버 한 대당 m명 수용)
        
        # 현재 시간대에 필요한 서버가 0 이상일 때만 처리
        if server > 0:
            # 만약 현재 시간대 운용 중인 서버가 부족하면 증설 필요
            if servers[x] < server:
                # 부족한 서버 수만큼 증설
                diff = server - servers[x]
                count += diff  # 증설 횟수 누적
                
                # 증설된 서버는 k시간 동안 운용 가능
                # 현재 시간대부터 x+k 시점까지 servers에 반영
                # 리스트 슬라이싱과 리스트 내포(list comprehension) 사용
                servers[x:x+k] = [u + diff for u in servers[x:x+k]]
    
    return count
```
