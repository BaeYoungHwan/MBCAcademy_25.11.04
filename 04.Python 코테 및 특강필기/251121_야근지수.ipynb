```python
# GPT가 알려준 힙버전 로직은 위가더 쉬울지 몰라도 효율성에서 차이가 많이남
# 이부분은 공부해서 파악해야함.
import heapq

def solution(n, works):
    # 최대 힙을 만들기 위해 음수로 변환
    max_heap = [-w for w in works]
    heapq.heapify(max_heap)

    # n시간 동안 반복
    for _ in range(n):
        if not max_heap:
            break
        # 가장 큰 작업량 꺼내기
        current = -heapq.heappop(max_heap)
        # 만약 0보다 크다면 1만큼 줄이고 다시 넣기
        if current > 0:
            heapq.heappush(max_heap, -(current - 1))

    # 남은 작업량이 양수인 것들을 제곱해서 더함
    fatigue = 0
    for w in max_heap:
        if w < 0:  # 음수로 저장했기 때문에 -w가 실제 작업량
            fatigue += (-w) ** 2

    return fatigue
```
