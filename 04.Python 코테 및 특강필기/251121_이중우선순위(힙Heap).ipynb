```python
operations = ["I -45", "I 653", "D 1", "I -642", "I 45", "I 97", "D 1", "D -1", "I 333"]
heap = []


while operations:

    operationPop = operations.pop(0)
    wordID, number = operationPop.split(" ")
    if wordID == "I":
        heap.append(int(number))
        
    elif wordID == "D":
        if heap:
            if number == "-1":
                heap.remove(min(heap))

            elif number == "1":
                heap.remove(max(heap))


                
        
print(heap)

if heap:
    return[max(heap),min(heap)]

else:
    return [0,0]
      



```
