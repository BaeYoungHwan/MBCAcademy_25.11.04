```python
import itertools

def get_score(guess, answer):
    """두 숫자를 비교해 스트라이크, 볼 개수를 계산하는 유틸리티 함수."""
    guess = str(guess)
    answer = str(answer)
    strike = sum(g == a for g, a in zip(guess, answer))
    ball = sum(g in answer for g in guess) - strike
    return f"{strike}S {ball}B"


def solution(n, submit):
    # 가능한 모든 4자리 조합 (1~9 중 서로 다른 4개)
    candidates = ["".join(map(str, comb)) for comb in itertools.permutations(range(1, 10), 4)]

    attempts = 0

    while candidates:
        # 가장 첫 번째 후보를 시도
        guess = candidates[0]
        result = submit(int(guess))
        attempts += 1

        # 정답이면 종료
        if result == "4S 0B":
            print(f"✅ 정답 {guess} (시도 {attempts}회)")
            return int(guess)

        # 스트라이크/볼 정보 파싱
        s, b = map(int, [result[0], result[3]])

        # 후보 갱신: 현재 guess와 같은 판정이 나올 수 있는 숫자만 남김
        new_candidates = []
        for cand in candidates:
            cand_score = get_score(cand, guess)
            cs, cb = map(int, [cand_score[0], cand_score[3]])
            if cs == s and cb == b:
                new_candidates.append(cand)

        candidates = new_candidates

    return -1  # 정답을 찾지 못한 경우 (이론상 발생 X)
```

```python
import java.util.function.Function;
import java.util.*;

class Solution {
    public int solution(int n, Function<Integer, String> submit) {
        List<Integer> list = new ArrayList<>();
        insertAllCase(list);

        while (list.size() != 1) {
            int num = list.get(0);
            String result = submit.apply(num);
            list = filterList(list, num, result.charAt(0) - '0', result.charAt(3) - '0');
        }

        return list.get(0);
    }

    // 3024개 전체 경우 생성
    private void insertAllCase(List<Integer> list) {
        for (int a = 1; a <= 9; a++) {
            for (int b = 1; b <= 9; b++) {
                if (a == b) continue;
                for (int c = 1; c <= 9; c++) {
                    if (a == c || b == c) continue;
                    for (int d = 1; d <= 9; d++) {
                        if (a == d || b == d || c == d) continue;
                        list.add(1000 * a + 100 * b + 10 * c + d);
                    }
                }
            }
        }
    }

    // submit 결과에 따른 list 업데이트
    private List<Integer> filterList(List<Integer> list, int num, int strike, int ball) {
        List<Integer> newList = new ArrayList<>();
        String strNum = Integer.toString(num);
        Set<Character> numSt = new HashSet<>();
        numSt.add(strNum.charAt(0));
        numSt.add(strNum.charAt(1));
        numSt.add(strNum.charAt(2));
        numSt.add(strNum.charAt(3));

        for (int ele : list) {
            String strEle = Integer.toString(ele);
            int strikeCnt = 0;
            int ballCnt = 0;

            for (int i = 0; i < 4; i++) {
                if (strNum.charAt(i) == strEle.charAt(i)) {
                    strikeCnt++;
                } else if (numSt.contains(strEle.charAt(i))) {
                    ballCnt++;
                }
            }

            if (strikeCnt == strike && ballCnt == ball) {
                newList.add(ele);
            }
        }

        return newList;
    }
}
```
