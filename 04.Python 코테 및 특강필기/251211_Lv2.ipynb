```python
#GpT의 정답지 나중에 보자


def solution(targets):
    # 1. 끝점 기준 오름차순 정렬
    targets.sort(key=lambda x: x[1])
    
    answer = 0
    end = -1  # 현재 요격 미사일이 설치된 위치
    
    # 2. 순회하면서 요격 위치 판단
    for s, e in targets:
        if s >= end:      # 기존 요격 위치로 막지 못하는 경우
            answer += 1    # 새 요격 미사일 발사
            end = e        # 끝나는 지점에 새로 설치
    
    return answer
```
