```python
def solution(s):
    score = 0
    for x in s:
        if x == ")":
            score -= 1
        else:
            score += 1
    
        if score < 0:
            
            return False

    score = 0        
    for x in  reversed(s):
        if x == "(":
            score -= 1
        else:
            score += 1
    
        if score < 0:
            
            return False
            
    
    return True

```

```python
def solution(progresses, speeds):
    resultList = []
    while progresses != []:
        score =0
        #100점을 도달 하지 못하면 돌아가는 반복문
        while progresses[0] < 100:
            for x in range(len(progresses)):
                progresses[x] = progresses[x] + speeds[x]
        #100점이 도달하면 리스트에서 빼고 스코어를 등록 중복되면 되는 만큼 스코어를 등록
        while progresses[0] >= 100:
           
            progresses.pop(0)
            speeds.pop(0)
            score += 1
            if progresses == []:
                break
        
        resultList.append(score)    

    return resultList

    
```
